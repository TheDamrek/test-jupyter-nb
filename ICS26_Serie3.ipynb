## Exercice 6 : docstrings 

Lorsqu'on définit une fonction, il est utile d'expliquer ce qu'elle fait, quels sont ses paramètres, et ce qu'elle retourne. On peut bien sûr utiliser des commentaires (lignes commençant par `#`), mais Python offre une convention plus formelle : les **docstrings**.

Une docstring est un `str` placé **immédiatement après la ligne** `def`, entouré de triple guillemets (`"""` ou `'''`). Elle est accessible via la fonction built-in `help`. Voici un exemple.

```python
def aire_triangle(base, hauteur):
    """
    Calcule l'aire d'un triangle.
    
    Paramètres:
        base    : longueur de la base (nombre positif)
        hauteur : hauteur du triangle (nombre positif)

    Retourne:
        Aire d'un triangle correspondant à base et hauteur
    """
    return base * hauteur / 2

help(aire_triangle)
```
