## Exercice 1 : countdown récursif

**(a)** Ecrivez une fonction **récursive** (sans boucle) `countdown` qui prend un entier positif `n` en entrée et affiche les nombres de `n` à `0` par ordre décroissant. Par exemple, `countdown(3)` doit afficher

```
3
2
1
0
```

N'oubliez pas de tester votre fonction pour quelques petites valeurs de `n`.

Que se passe-t-il si vous appelez `countdown` avec l'argument `n = -1` ?

**(b)** Ecrivez une fonction itérative `countdown_iter` (qui peut contenir des boucles mais pas d'appels récursifs) qui produit le même affichage.

````python
def countdown(n):
    if n == 0:
        return n
    
    print(n)
    return(countdown(n - 1))

print(countdown(5))
    
````

````python
def countdown_iter(n):
    for i in range(n):
        print (n - i)
    return 0

print(countdown_iter(5))
````

## Exercice 2 : maximum récursif

Ecrivez une fonction **récursive** (sans boucle) `max_rec` qui prend en entrée une liste `L` non vide de nombres et retourne le maximum de `L`.

N'oubliez pas de tester votre fonction sur quelques exemples de `L`.

Que se passe-t-il si vous appelez votre fonction avec une liste vide ?

**Remarque**: à l'exercice 5 de la série 4, vous avez déjà écrit une fonction itérative (avec boucle, sans appels récursifs) qui fait ce même calcul du maximum d'une liste `L`, sans appeler la fonction `max()`. Sauriez-vous la réécrire?

````python
def max_rec(L):
    if len(L) == 1:
        return L[0]
    
    max_ = max_rec(L[1:])
    
    if L[0] > max_:
        max_ = L[0]
        return max_
    else:
        return max_
    
print(max_rec([0, 90, 34, -9, 23]))
````

## Exercice 3 : factorielle et Fibonacci itératifs

Vous avez vu au cours des fonctions récursives pour calculer la factorielle d'un nombre `n` et le n-ième nombre de Fibonacci. Dans cet exercice, vous allez implémenter des versions itératives (avec une boucle) de ces fonctions.

**(a)** Ecrivez une fonction `fact_iter` qui prend en entrée un entier $n \geq 0$ et calcule itérativement (avec une boucle, sans appels récursifs) la factorielle de `n`.

**(b)** Ecrivez une fonction `fib_iter` qui prend en entrée un entier $n \geq 0$ et calcule itérativement (avec une boucle, sans appels récursifs) le n-ième nombre de Fibonacci.

*Indice :* A travers les itérations de la boucle, vous devez vous rappeler de deux valeurs : les deux derniers nombres de Fibonacci calculés.


````python
def fact_iter(n):
    result = 1
    if n == 0:
        return 1
    for i in range(1, n + 1):
        result = result * i
    return result
        
print(fact_iter(5))      
````

````python
def fib_iter(n):
    if n == 0:
        return 0
    a = 0
    b = 1
    
    for i in range(n):
        if 1 < i < n:
            a, b = b, a + b
    return a + b
        

        
print(fib_iter(10))    
````

## Exercice 4 : reverse récursif

Ecrivez une fonction récursive `reverse_rec` qui prend en
entrée une chaîne de caractères `s` et retourne la chaîne de caractères
constituée des mêmes caractères que `s` mais dans l'ordre inverse.

Par exemple, `reverse_rec("abc")` doit retourner `"cba"`,
`reverse_rec("a")` doit retourner `"a"` et
`reverse_rec("")` doit retourner la chaîne vide.

Votre fonction ne doit contenir aucune boucle. Elle ne doit pas
utiliser de slicing de pas négatif (`s[::-1]`), ni la fonction
`reversed()`, ni la méthode de liste `reverse()`.

**Remarque**: à l'exercice 4 de la série 4, vous avez déjà écrit une fonction itérative (avec boucle, sans appels récursifs) qui fait un calcul similaire, mais sur une liste `L`. Sauriez-vous la réécrire (en l'adaptant pour une chaîne de caractères)?

````python
def reverse_rec(str):
    if str == "":
        return ""
    
    return str[-1] + reverse_rec(str[:-1])
    
print(reverse_rec("Thanks"))
````
