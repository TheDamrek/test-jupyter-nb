## Exercice 2 : division euclidienne

Pour des entiers $n\geq 0$ et $d > 0$, il existe d'uniques entiers naturels $q$ et $r$, avec $0 \leq r < d$, tels que

$$n = q \cdot d + r.$$

$q$ est le quotient et $r$ le reste de la division euclidienne de $n$ par $d$.

Par exemple,
- comme $13 = 3 \times 4 + 1$, le quotient et le reste de la division euclidienne de $13$ par $3$ sont respectivement $4$ et $1$...
- ... et le quotient et le reste de la division euclidienne de $13$ par $4$ sont respectivement $3$ et $1$.
- comme $29 = 2 \times 10 + 9$, le quotient et le reste de la division euclidienne de $29$ par $10$ sont respectivement $2$ et $9$. 
- par contre l'égalité ci-dessus n'implique pas que le quotient et le reste de la division euclidienne de $29$ par $2$ sont respectivement $10$ et $9$! N'oubliez pas que le reste de la division entière doit être strictement inférieur au diviseur. Quels sont le quotient et le reste de la division entière de $29$ par $2$?

En Python, l'opérateur `//` donne le quotient de la division entière, et `%` le reste de la division entière:

```python
print("le quotient de la division entière de 13 par 4 est", 13 // 4, "et le reste est", 13 % 4)
print("le quotient de la division entière de 13 par 3 est", 13 // 3, "et le reste est", 13 % 3)
print("le quotient de la division entière de 29 par 10 est", 29 // 10, "et le reste est", 29 % 10)
print("le quotient de la division entière de 29 par 2 est", 29 // 2, "et le reste est", 29 % 2)
```

<font color = grey>Notez que les opérateurs `//` et `%` sont définis également pour des entiers négatifs ou même des float, mais ces cas présentent moins d'intérêt mathématique.
<br/><br/><br/></font>

Ecrivez un programme qui
- définit une variable `n` et une variable `d` (et leur affecte des entiers positifs aléatoires ou de votre choix)
- affecte à deux variables `q` et `r` le quotient et le reste de la division euclidienne de `n` par `d` 
- affiche l'équation 

`n = d * q + r`


Par exemple, si `n` vaut `11` et `d` vaut `4`, le programme doit afficher:

`11 = 4 * 2 + 3`

Testez votre programme pour des entiers positifs aléatoires ou pour des valeurs de votre choix pour `n` et de `d`, par exemple:

- `n = 129, d = 4`
- `n = 33, d = 7`
- `n = 7, d = 33`
- `n = 13, d = 13`



```python
from random import randint

# modifiez ces lignes de code pour changer la plage des nombres aléatoires, ou affecter les valeurs de votre choix à n et d
n = randint(50, 100)
d = randint(10, 40)

q = n // d
r = n % d

print(q, r)
print("n = d * q + r")
print("n =", d * q + r)
print("n =", int(d) * int(q) + int(r))
print("n =", int(d), "*", int(q), "+", int(r))


# complétez le code

```

## Exercice 3 : signe d'un entier

Ecrivez un programme qui génère un entier aléatoire entre -100 et 100, et affiche cet entier en précisant s'il est négatif, nul ou positif.

**Remarque 1:** On ne sait pas encore tester l'égalité (ce n'est pas avec un signe = !). Mais vous n'avez pas besoin de tester explicitement une égalité. 

**Remarque 2:** Faites tourner votre code plusieurs fois pour tester tous les cas de figure. Si vous voulez « forcer » un certain cas de test, par exemple le cas où l'entier est nul, vous pouvez aussi tout simplement donner à l'entier la valeur que vous voulez au lieu d'une valeur aléatoire.


```python
n = randint(-100, 100)

print(n)

if n < 0:
    print("n est négatif")
elif n == 0:
    print("n est nul")
else:
    print("n est positif")
```

## Exercice 5: classification d'un nombre aléatoire

Ecrivez un programme qui définit une variable `x` et lui affecte un entier aléatoire entre 0 et 200, puis selon la valeur de `x`, produit
un des affichages suivants:

`0 <= x < 50`

`50 <= x < 100`

`100 <= x <= 200`

(avec `x` remplacé par sa valeur dans l'affichage.)
