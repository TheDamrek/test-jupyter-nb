## Exercice 3 : somme des entiers pairs

**(a)** Ecrivez un programme qui définit une variable `n` et lui affecte une valeur entière positive aléatoire ou de votre choix, puis calcule et affiche la somme des entiers pairs entre `0` et `n` inclus, à l'aide d'une boucle `while`.

```python
from random import randint
n = randint(1, 100)

i = 0
somme = 0

while i < n:
    i += 2
    if i > n:
        break
    somme += i

print(somme)
```

**(b)** Même question mais avec une boucle `for`.

```python
from random import randint
n = randint(1, 100)

i = 0
somme = 0

for i in range(0, n, 2):
    if i > n:
        break
    somme +=i
print(somme)

```

**(c)** Même question, mais maintenant on permet à `n` d'être un entier relatif quelconque. Modifiez vos programmes des points (a) et (b) pour qu'ils calculent la somme des entiers pairs de `0` à `n` ou de `n` à `0`, selon le signe de `n`.

```python
import sys

from random import randint
n = randint(-sys.maxsize - 1, sys.maxsize - 1)

i = 0
somme = 0

if i < n:
    while i < n:
        i += 2
        if i > n:
            break
        somme += i

elif i > n:
    while i > n:
        i -= 2
        if i < n:
            break
        somme += i

print(somme)
```
