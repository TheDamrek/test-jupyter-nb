## Exercice 3


Ecrivez une fonction qui prend en entrée un entier `n` positif et qui retourne la liste contenant les nombres `1, 2, ... n`:

**(a)** En utilisant `list()`

**(b)** En utilisant une boucle 



```python
from random import randint

n = randint(1, 100)

L = list(range(1, n))

print("n =", n)
print(L)
```

```python
from random import randint

n = randint(1, 100)

L = list()

for i in range (1, n + 1):
    if i < n + 1:
        L.append(i)
        print("n =", n)
        print(L)
```

```python
from random import randint

n = randint(1, 100)

L = list()

for i in range (1, n + 1):
    if i < n + 1:
        L.append(i)

print(n)
print(L)
```
