```python
def de1An(n):
    return [x for x in range(1, n+1)]

print(de1An(17))
```

```python
def OnlyPair(L):
    return [x for x in L if x % 2 == 0]

print(OnlyPair([1, 76, 90, 43]))
```

```python
def list_intersection(L1, L2):
    return [x for x in L1 if x in L2]

print(list_intersection([21, 76, 90,  87, 55, 1, 66, 93, 20, 5], [21, 78, 56, 0, 23, 66, 54, 31]))
```

## Exercice 7 : addition de matrices

Une **matrice** de dimension $m \times n$ est un objet mathématique qui peut être informatiquement représenté par un tableau à $m$ lignes et $n$ colonnes.

On peut définir l'opération d'**addition** sur deux matrices de dimensions $2 \times 2$:

$\begin{pmatrix}
a_1 & b_1\\
c_1 & d_1
\end{pmatrix} + 
\begin{pmatrix}
a_2 & b_2\\
c_2 & d_2
\end{pmatrix} =
\begin{pmatrix}
a_1+a_2 & b_1+b_2\\
c_1+c_2 & d_1+d_2
\end{pmatrix}$

En général, la somme de deux matrices $A$ et $B$ de dimensions  $m \times n$ sera une matrice de dimensions $m \times n$ où l'élément de coordonnées $(i, j)$ (c'est-à-dire l'élément à la ligne $i$ et colonne $j$) est la somme des éléments de coordonnées $(i, j)$ dans chacune des deux matrices $A$ et $B$. Par exemple

$\begin{pmatrix}
1 & 2 & 3\\
4 & 5 & 6\\
\end{pmatrix} + 
\begin{pmatrix}
10 & 20 & 30\\
40 & 50 & 60\\
\end{pmatrix} =
\begin{pmatrix}
11 & 22 & 33\\
44 & 55 & 66\\
\end{pmatrix}.$

**(a)** Ecrivez une fonction `somme_mat_2x2` qui prend en entrée deux tableaux représentant des matrices $2 \times 2$, comme vu au cours, et qui retourne un tableau représentant la somme des deux matrices données.

Par exemple, l'appel `somme_mat_2x2([[1, 2],[3, 4]], [[5, 6],[7, 8]])` doit retourner `[[6, 8],[10, 12]]`.
