## Exercice 2 : compréhensions de listes 

Voici quelques fonctions que vous avez implémentées à la série 4 à l'aide de boucles `for` et de la méthode `append`. Cette fois, implémentez-les avec des compréhensions de listes.

**(a)** Ecrivez une fonction qui prend en entrée un entier `n` positif et qui retourne la liste contenant les nombres `1, 2, ... n`.

**(b)** Ecrivez une fonction qui prend en entrée une liste `L` de nombres et retourne une liste `L1` contenant uniquement les éléments pairs de `L`.

**(c)** Ecrivez une fonction `list_intersection` qui prend en entrée deux listes `L1` et `L2` et retourne une liste `L` contenant l'intersection de `L1` et `L2`, c'est-à-dire les éléments présents dans les deux listes. Par exemple, l'appel `list_intersection([1, 3, 5], [2, 3])` doit retourner la liste `[3]`.

```python
# (a)

def liste_nombres(n):
    return [x for x in range(1, n+1)]

print(liste_nombres(8))
```

```python
# (b)

def liste_pairs(L):
    return [x for x in L if x % 2 == 0]

print(liste_pairs([1, 2, 3, 2]))
print(liste_pairs([1, 3]))
```

```python
# (c)

def list_intersection(L1, L2):
    return [x for x in L1 if x in L2]

print(list_intersection([1, 3, 5], [2, 3]))
print(list_intersection([1, 3, 5], [2, 4]))
```

## Exercice 4 : calculs sur les valeurs d'un dictionnaire

**(a)** Ecrivez une fonction qui prend en entrée un dictionnaire `d` dont les valeurs sont numériques, et retourne la clé associée à la valeur maximale (si plusieurs clés sont associées à la valeur maximale, vous pouvez retourner n'importe laquelle de ces clés). Si le dictionnaire est vide, votre fonction doit retourner `None`. 
Ecrivez deux versions de la fonction:
1. en utilisant la fonction `max()`
2. sans utiliser la fonction `max()` (*Indice*: il sera utile d'initialiser une certaine variable à `float("-inf")`, qui se comporte comme $-\infty$ (en particulier, elle est plus petite que toute autre valeur numérique)).

**(b)** Ecrivez une fonction qui prend en entrée un dictionnaire `d` dont les valeurs sont numériques, et retourne la somme des valeurs du dictionnaire. Si le dictionnaire est vide, votre fonction doit retourner `0`.


```python
# (a)

def dict_max1(d):
    if d: # d est non vide
        m = max(d.values())
        for c in d:
            if d[c] == m:
                return c #retourne la première clé de valeur m
    
    # si d est vide on ne rentre pas dans le if et la fonction retourne None
    
def dict_max2(d):
    if d: # d est non vide
        d_max = float("-inf")
        cle_max = None 
        
        #cette boucle trouve la valeur maximale du dictionnaire et la clé associée
        for k, v in d.items():
            if v > d_max:
                d_max = v
                cle_max = k
                
        return cle_max

print(dict_max1({}))
print(dict_max2({}))
print(dict_max1({"a":1, "b":10, "c":-2, "d":10}))
print(dict_max2({"a":1, "b":10, "c":-2, "d":10}))
```

```python
# (b)

def dict_sum(d):
    s = 0
    for x in d:
        s += d[x]
    return s

print(dict_sum({"a":1, "b":10, "c":-2}))
```

## Exercice 5 : "inverser" un dictionnaire

**(a)** Ecrivez une fonction `dict_inverse` qui prend en entrée un dictionnaire `d` et retourne le dictionnaire dont les clés sont les **valeurs** de `d` et les valeurs associées sont les **clés** de `d`. Vous pouvez supposer que les valeurs du dictionnaire `d` sont uniques et non mutables.

Par exemple, l'appel `dict_inverse({1:"a", 2:"b"})` doit retourner le dictionnaire `{"a":1, "b":2}`.

Donnez 
1. Une version de la fonction qui utilise une boucle
2. Une version qui utilise la compréhension de dictionnaire.

```python
# avec une boucle

def dict_inverse(d):
    d1 = {}
    for x in d:
        d1[d[x]] = x
    return d1

d = {1:"a", 2:"b"}
d2 = dict_inverse(d)
print(d2)


# OU BIEN:

def dict_inverse2(d):
    d1 = {}
    for k, v in d.items():
        d1[v] = k
    return d1

d = {1:"a", 2:"b"}
d2 = dict_inverse2(d)
print(d2)
```

```python
# avec une compréhension 

def dict_inverse_compr(d):
    return {d[x]:x for x in d}

# OU BIEN:
def dict_inverse_compr2(d):
    return {v:k for k, v in d.items()}

d = {1:"a", 2:"b"}
print(dict_inverse_compr(d))
print(dict_inverse_compr2(d))
```

**(b)** Ecrivez une fonction `dict_inverse_list` qui prend en entrée un dictionnaire `d` dont les valeurs ne sont pas nécessairement uniques (mais toujours non mutables), et retourne le dictionnaire dont les clés sont les valeurs de `d`, et les valeurs associées sont les **listes** des clés de `d` correspondantes.

Par exemple, l'appel `dict_inverse_list({1:"a", 2:"b", 3:"a", 4:"c"})` doit retourner le dictionnaire `{"a":[1, 3], "b":[2], "c":[4]}`.


```python
def dict_inverse_list(d):

    d1 = {v:[] for v in d.values()}
    
    for x in d:
        d1[d[x]].append(x)
        
    return d1

# OU BIEN:
def dict_inverse_list2(d):

    d1 = {v:[] for v in d.values()}
    
    for k, v in d.items():
        d1[v].append(k)
        
    return d1

d = {1:"a", 2:"b", 3:"a", 4:"c"}
print(dict_inverse_list(d))
print(dict_inverse_list2(d))

```

## Exercice 6 : dictionnaire de fréquences

**(a)** Ecrivez une fonction `dict_occurrences` qui prend en entrée une liste `L` d'objets non mutables, et retourne le dictionnaire dont les clés sont les éléments de `L`, et la valeur associée à chaque clé `x` est le nombre de fois où `x` apparaît dans `L`.

Par exemple, `dict_occurrences(["a", "b", "a", "c", "a", "b"])` doit retourner le dictionnaire `{"a":3, "b":2, "c":1}`.


```python
def dict_occurrences(L):
    
    d = {}
    for x in L:
        d[x] = L.count(x)
        
    return d

L = [2, 6, 7, 8, 9, 8, 7, 6, 7]
print(dict_occurrences(L))

# la solution ci-dessous est plus rapide pour de grandes listes.
# on pourra mieux quantifier l'efficacité de chacune des deux solutions
# lorsqu'on abordera le chapitre de la complexité.

def dict_occurrences2(L):
    d = {x:0 for x in L}
    for x in L:
        d[x] += 1
      
    return d

L = [2, 6, 7, 8, 9, 8, 7, 6, 7]
print(dict_occurrences2(L))
```

**(b)** Ecrivez une fonction qui

- retourne le nombre d'éléments de `d` qui ont une valeur plus grande que `1`.

Testez votre code pour plusieurs choix des paramètres `n` et `m` (par exemple, `n = 1000` et `m = 10000`; `n = 10000` et `m = 1000`, `n = m = 1000`...) Pour chaque choix de `n` et `m`, lancez le code plusieurs fois et observez les résultats.

```python
import random

def dict_random_list(n, m):
    L = []
    for i in range(n):
        L.append(random.randint(1,m))
    
    d = dict_occurrences(L)

    count = 0
    
    for x in d:
        if d[x] > 1:
            count += 1
         
    return count

print(dict_random_list(100, 1000))
```

## Exercice 7 : addition de matrices

Une **matrice** de dimension $m \times n$ est un objet mathématique qui peut être informatiquement représenté par un tableau à $m$ lignes et $n$ colonnes.

On peut définir l'opération d'**addition** sur deux matrices de dimensions $2 \times 2$:

$\begin{pmatrix}
a_1 & b_1\\
c_1 & d_1
\end{pmatrix} + 
\begin{pmatrix}
a_2 & b_2\\
c_2 & d_2
\end{pmatrix} =
\begin{pmatrix}
a_1+a_2 & b_1+b_2\\
c_1+c_2 & d_1+d_2
\end{pmatrix}$

En général, la somme de deux matrices $A$ et $B$ de dimensions  $m \times n$ sera une matrice de dimensions $m \times n$ où l'élément de coordonnées $(i, j)$ (c'est-à-dire l'élément à la ligne $i$ et colonne $j$) est la somme des éléments de coordonnées $(i, j)$ dans chacune des deux matrices $A$ et $B$. Par exemple

$\begin{pmatrix}
1 & 2 & 3\\
4 & 5 & 6\\
\end{pmatrix} + 
\begin{pmatrix}
10 & 20 & 30\\
40 & 50 & 60\\
\end{pmatrix} =
\begin{pmatrix}
11 & 22 & 33\\
44 & 55 & 66\\
\end{pmatrix}.$

**(a)** Ecrivez une fonction `somme_mat_2x2` qui prend en entrée deux tableaux représentant des matrices $2 \times 2$, comme vu au cours, et qui retourne un tableau représentant la somme des deux matrices données.

Par exemple, l'appel `somme_mat_2x2([[1, 2],[3, 4]], [[5, 6],[7, 8]])` doit retourner `[[6, 8],[10, 12]]`.

**(b)** Ecrivez une fonction `somme_mat` qui prend en entrées deux tableaux représentant deux matrices de même taille, et qui retourne un tableau représentant la somme des deux matrices données.
Remarquez qu'à la différence de la partie (a) (où vous avez pu construire la matrice somme "à la main"), il vous faudra maintenant itérer sur les listes données. Commencez par trouver les dimensions des matrices...

```python
#(a)

def somme_mat_2x2(A, B):
    return [ [A[0][0] + B[0][0], A[0][1] + B[0][1]], [A[1][0] + B[1][0], A[1][1] + B[1][1]] ]

L1 = [[1, 2],[3, 4]]
L2 = [[5, 6],[7, 8]]

print(somme_mat_2x2(L1, L2))
```

```python
# (b)

def somme_mat(A, B):
    '''
    Attention, si vous me donnez A et B de dimensions différentes, je ne garantis rien!
    '''
    m = len(A) # nombre de lignes dans les matrices
    n = len(A[0]) # nombre de colonnes
    C = []
    for i in range(m): # on itère sur chaque ligne
        C.append([]) #la ième ligne, initialement vide
        for j in range(n): # on itère sur chaque colonne
            C[i].append(A[i][j] + B[i][j]) # on ajoute à la ième ligne l'élément de la ligne i et colonne j
    return C

L1 = [[1, 2], [0, 0], [3, -1]]
L2 = [[5, 6], [10, 11], [-1, -2]]  
print(somme_mat(L1, L2))
    
```
